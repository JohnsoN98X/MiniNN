```python
import numpy as np
from mini_nn.layers import ReLu
```

```python
relu = ReLu()
type(relu)
```

# Check the Forward Propagation

```python
forward_input = np.array([-1, 1, -2, 0, 3]).astype(float)
output = relu(forward_input)

print(f'input shape: {forward_input.shape}')
print(f'forward output shape: {output.shape}')
print()
print(f'input: {forward_input}')
print(f'output: {output}')
```

```python
# 2D check:

x = np.array([
    [-1, 2, -3],
    [4, -5, 6]
])

output = relu(x)
print(output)
print(output.shape)
```

# Backward Check

```python
backward_inputs = x + np.sin(2)
relu.backward(backward_inputs)
```
