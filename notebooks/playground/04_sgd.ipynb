```python
from mini_nn.optim import SGD
from mini_nn.layers import Linear, ReLu
import numpy as np
```

```python
linear1 = Linear(3,2)
relu = ReLu()
linear2 = Linear(2,1)

optimizer = SGD(
    layers=[linear1, relu, linear2],
    learning_rate=0.001
)
```

```python
linear1.dw = np.ones_like(linear1.weight)
linear1.db = np.ones_like(linear1.bias)

linear2.dw = np.full(linear2.weight.shape, fill_value=2)
linear2.db = np.ones_like(linear2.bias)

linear1_weights_before = linear1.weight.copy()
linear1_bias_before = linear1.bias.copy()

linear2_weights_before = linear2.weight.copy()
linear2_bias_before = linear2.bias.copy()
```

```python
optimizer.step()
```

```python
print(f'linear1 weights before:\n{linear1_weights_before}')
print(f'linear1 weights after:\n{linear1.weight}')
print('='*30)
print(f'linear bias before:\n{linear1_bias_before}')
print(f'linear bias after:\n{linear1.bias}')
```

```python
print(f'linear2 weights before:\n{linear2_weights_before}')
print(f'linear2 weights after:\n{linear2.weight}')
print('='*30)
print(f'linear2 bias before:\n{linear2_bias_before}')
print(f'linear2 bias after:\n{linear2.bias}')
```
