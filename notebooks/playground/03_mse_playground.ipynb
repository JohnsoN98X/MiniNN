```python
import numpy as np
from mini_nn.losses import MSELoss
```

```python
loss_fn = MSELoss()
```

# Forward Check

```python
gen = np.random.default_rng(11)
y_true = gen.integers(1, 10, 50)
y_pred = gen.integers(1.5, 12, 50)
```

```python
mse = loss_fn(y_true, y_pred)
manual_mse = np.mean((y_pred - y_true)**2)
print(f'loss function mse value: {mse}')
print(f'manual calc mse value: {manual_mse}')
```

# Backward Check

```python
grad_pred = loss_fn.backward()
expected_grad = (2 / y_pred.size) * (y_pred - y_true)

print(f"backward check results: {'Positive' if {np.allclose(grad_pred, expected_grad)} else 'negative'} ")
```
