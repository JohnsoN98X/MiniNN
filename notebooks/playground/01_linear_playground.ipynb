```python
import numpy as np
from mini_nn.layers import Linear
```

```python
linear_layer = Linear(in_features=3, out_features=2)
```

# Properties and Attributes Check

```python
print(f'linear layer input features: {linear_layer.in_features}')
print(f'linear layer output features: {linear_layer.out_features}')
```

```python
print(f'linear layer wegihts shape: {linear_layer.weight.shape}')
print(f'linear layer total weights count: {linear_layer.weight.size}')
print(f'linear layer random weights (initialization stage): \n{linear_layer.weight}')
```

```python
print(f'linear layer bias shape: {linear_layer.bias.shape}')
print(f'linear layer total bias count: {linear_layer.bias.size}')
print(f'linear layer bias (initialization stage): \n{linear_layer.bias}')
```

# Forward Pass Check

```python
X = np.array([[1,3,5],
              [2,4,6]]).astype(float)
```

```python
Y = linear_layer.forward(X)

print(f'X (input) shape: {X.shape}')
print(f'Y (output) shape: {Y.shape}')
print('Y:')
print(Y)
```

```python
manual_y = X @ linear_layer.weight + linear_layer.bias
print(f'Manual Y0:\n{manual_y0}')
print(f'Similatiry numerical check passed: {np.allclose(manual_y, Y)}'
)
```

#

# Backpropagation Check

```python
grad_outputs = np.ones_like(Y)

grad_inputs = linear_layer.backward(grad_outputs)

print(f'grad_outputs shape: {grad_outputs.shape}')
print(f'dw shape: {linear_layer.dw.shape}')
print(f'db shape: {linear_layer.db.shape}')
print(f'grad inputs shape: {grad_inputs.shape}')
```

```python
manual_dw = X.T @ grad_outputs
manual_db = np.sum(grad_outputs, axis=0, keepdims=True)
manual_grad_inputs = grad_outputs @ linear_layer.weight.T

print("dw equal:", np.allclose(linear_layer.dw, manual_dw))
print("db equal:", np.allclose(linear_layer.db, manual_db))
print("grad_inputs equal:", np.allclose(grad_inputs, manual_grad_inputs))
```
